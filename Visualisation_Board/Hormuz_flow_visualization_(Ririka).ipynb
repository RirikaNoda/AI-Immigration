```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/Users/riritorti/Documents/GitHub/AI-Immigration/Data/raw/crude_oil_export_voy_intake_index.csv")
```

```python
print(df.head())
```

```python
print(df.columns)
```

```python
print(df.info())
```

```python
df['date'] = pd.to_datetime(df['date'])
print(df.dtypes)
```

```python
df_2026 = df[df['date'].dt.year == 2026].copy()
print(df_2026.head())
```

```python
df_2026['ma_7day'] = (
    df_2026['voy_intake_index']
    .rolling(window=7)
    .mean()
)
```

```python
print(df_2026[['date', 'voy_intake_index', 'ma_7day']].head(15))
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    df_2026['date'],
    df_2026['voy_intake_index']
)

plt.title("Daily Voyage Intake Index (2026)")
plt.xlabel("Date")
plt.ylabel("Index Value")

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    df_2026['date'],
    df_2026['voy_intake_index'],
    alpha=0.5,
    label='Daily Index'
)

plt.plot(
    df_2026['date'],
    df_2026['ma_7day'],
    linewidth=2,
    label='7-Day Moving Average'
)

plt.legend()

plt.title("Crude Oil Export through the Strait of Hormuz (2026)")
plt.xlabel("Date")
plt.ylabel("Index Value")

plt.show()
```

```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(12,6))

sns.lineplot(
    data=df_2026,
    x='date',
    y='voy_intake_index',
    label='Daily Voyage Intake Index',
    alpha=0.5
)

sns.lineplot(
    data=df_2026,
    x='date',
    y='ma_7day',
    label='7-Day Moving Average'
)

plt.title('Crude Oil Export through the Strait of Hormuz (2026)')
plt.xlabel('Date')
plt.ylabel('Index Value')

plt.show()
```
